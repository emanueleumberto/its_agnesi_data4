Crea una classe Libro con:
attributi: titolo, autore, pagine
metodo descrizione() che restituisce una stringa descrittiva del libro
Istanzia almeno 2 oggetti e stampa la loro descrizione

Vincoli:
Usare __init__
Non usare librerie esterne

Output atteso (esempio):
"Il libro '1984' di George Orwell ha 328 pagine"

```python
class Libro:
    def __init__(self, titolo, autore, pagine):
        self.titolo = titolo
        self.autore = autore
        self.pagine = pagine
        
    def descrizione(self):
        return f"Il libro '{self.titolo}' di {self.autore} ha {self.pagine} pagine"
    
libro1 = Libro('1984', 'George Orwell', 328)
libro2 = Libro("Il Signore degli Anelli", "J.R.R. Tolkien", 1200)

print(libro1.descrizione())
print(libro2.descrizione())
```

Aggiungere il tipo di supporto (Digitale o Cartaceo)
Aggiungere il prezzo in formato decimale

Creare 5 Libri, e inserirli in una lista e calcolare prezzo medio, prezzo più alto e prezzo più basso. E dividiamo i libri in set separati per Digitale o Cartaceo

```python
from decimal import Decimal

class Libro:
    
    __contatore = 0
    
    def __init__(self, titolo, autore, pagine, supporto, prezzo):
        self.__titolo = titolo
        self.__autore = autore
        self.__pagine = pagine
        self.__supporto = self.valida_supporto(supporto)
        self.__prezzo = prezzo
        Libro.__contatore += 1
        
        
    @property 
    def titolo(self): return self.__titolo
    
    @property 
    def autore(self): return self.__autore
    
    @property 
    def pagine(self): return self.__pagine
    
    @property
    def supporto(self): return self.__supporto
    
    @property
    def prezzo(self): return self.__prezzo
    
    @supporto.setter
    def supporto(self, supporto): 
        self.__supporto = self.valida_supporto(supporto)
      
    @prezzo.setter
    def prezzo(self, prezzo):
        if prezzo < 0:
              self.__prezzo = Decimal(0) 
        self.__prezzo = Decimal(prezzo) 
          
    def valida_supporto(self, supporto):
        if supporto.capitalize() == 'Digitale':
            return 'Digitale'
        elif supporto.capitalize() == 'Cartaceo':
            return 'Cartaceo'
        else:
            print('Hai inserito un supporto non valido di default verrà inserito Cartaceo')
            return 'Cartaceo'
 
 
    def descrizione(self):
        return f"Il libro '{self.titolo}' di {self.autore} ha {self.pagine} pagine"
    
    
libri = [
    Libro("1984", "Orwell", 328, "Cartaceo", 12.50),
    Libro("Clean Code", "Martin", 450, "Cartaceo", 30.00),
    Libro("Python Crash Course", "Matthes", 544, "Digitale", 25.99),
    Libro("Il Signore degli Anelli", "Tolkien", 1200, "Cartaceo", 45.00),
    Libro("Automate the Boring Stuff", "Sweigart", 400, "Digitale", 19.99)
]

for l in libri: print(l.titolo)
print("###########")

prezzi = [l.prezzo for l in libri]
media = sum(prezzi)/len(libri)
print(f"La media del prezzo dei libri è: {media}")
print(f"Il prezzo più alto dei libri nella lista è: {max(prezzi)}")
print(f"Il prezzo più basso dei libri nella lista è: {min(prezzi)}")
print("###########")
digitali = {l for l in libri if l.supporto == 'Digitale'}
cartaceo = {l for l in libri if l.supporto == 'Cartaceo'}
for l in digitali: print(l.titolo)
print("###########")
for l in cartaceo: print(l.titolo)
```
