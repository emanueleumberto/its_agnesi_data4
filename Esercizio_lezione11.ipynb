Sistema di gestione di un conto bancario (uso di classi, incapsulamento e metodi dunder)
Progettare una classe ContoBancario che rappresenti un conto corrente con le seguenti caratteristiche:
-> Attributi (incapsulati):
    __titolare (stringa)
    __saldo (float)
    __movimenti (lista di stringhe che descrivono le operazioni)

-> Metodi pubblici:
    deposita(importo)
        Aggiunge denaro al saldo
        Registra il movimento
        Deve validare che l’importo sia positivo
    preleva(importo)
        Sottrae denaro dal saldo
        Impedisce saldo negativo
        Registra il movimento
    get_saldo()
        Restituisce il saldo attuale
    get_movimenti()
        Restituisce la lista dei movimenti

-> Requisiti sui metodi dunder:
    __init__(self, titolare, saldo_iniziale)
        Inizializza il conto
        Se il saldo iniziale è negativo → impostare a 0
    __str__(self)
        Restituisce una stringa leggibile:
        Es: "Conto di Mario Rossi - Saldo: 1500€"
    __repr__(self)
        Restituisce una rappresentazione tecnica utile al debug
    __eq__(self, altro)
        Due conti sono uguali se:
            Hanno lo stesso titolare
            Hanno lo stesso saldo
    __len__(self)
        Restituisce il numero di movimenti effettuati

Output atteso (indicativo):
- Saldo aggiornato correttamente
- Movimenti registrati
- Confronto tra oggetti funzionante
- Rappresentazione leggibile e tecnica separate

```python
class ContoBancario:
    
    def __init__(self, titolare, saldo_iniziale):
        self.__titolare = titolare
        self.__saldo = saldo_iniziale if saldo_iniziale >= 0 else 0.0
        self.__movimenti = []
    
    @property
    def titolare(self): return self.__titolare
     
    @property
    def saldo(self): return self.__saldo
    
    @property
    def movimenti(self): return self.__movimenti
    
    def __str__(self):
        return f"Conto di {self.titolare} - Saldo: {self.saldo}€"
    
    def __repr__(self):
        return f"ContoCorrente(Conto di {self.titolare} - Saldo: {self.saldo}€ - {len(self.movimenti)}) - {self.movimenti}"
    
    def __eq__(self, other):
        if not isinstance(other, ContoBancario):
            print('Non puoi confrontare oggetti diversi')
            return False
        return self.titolare == other.titolare and self.saldo == other.saldo
    
    def __len__(self):
        return len(self.movimenti)
        
    def deposita(self, importo):
        if importo <= 0:
            print('Importo non valido. Deve essere positivo')
            return
        self.__saldo += importo
        self.__movimenti.append(f'Deposito di €{importo} - Saldo:{self.saldo}€')
        
    def preleva(self, importo):
        if importo <= 0:
            print('Importo non valido. Deve essere positivo')
            return
        if self.__saldo < importo:
            print('Saldo insufficiente')
            return
        self.__saldo -= importo
        self.__movimenti.append(f'Prelievo di €{importo} - Saldo:{self.saldo}€')
        
        
########################################

conto1 = ContoBancario('Mario Rossi', 100)
print(conto1)

conto2 = ContoBancario('Giuseppe Verdi', -50)
print(conto2)

conto1.deposita(500.0)
conto1.preleva(60.0)
print(repr(conto1))

print('Conto1 è uguale a Conto2?', conto1 == conto2)

print(f'Movimenti registrati sul Conto1 di {conto1.titolare}')
# for movimento in conto1.movimenti:
#     print(f'  - {movimento}')

print([f'  - {movimento}' for movimento in conto1.movimenti])
```

Sistema di gestione di una libreria digitale (catalogo libri)

Progettare un sistema composto da due classi:
- Libro
- Libreria

Il sistema deve permettere la gestione di un catalogo libri con operazioni di inserimento, ricerca e analisi.

Requisiti classe Libro:
Attributi (privati):

Requisiti classe Libreria:

Metodi pubblici:

    rimuovi_libro(isbn)

    libri_disponibili()
    totale_pagine()
        Somma delle pagine di tutti i libri

Metodi dunder richiesti:

Per Libro:

Per Libreria:
