Ereditarietà
Crea una classe base Veicolo con:
attributi: marca, modello
metodo descrizione()
Crea due classi derivate:
Auto → aggiunge porte
Moto → aggiunge cilindrata
Override del metodo descrizione() nelle classi figlie

Vincoli:
Usare super()
Evitare duplicazione di codice

Output atteso:
"Auto Fiat Panda con 5 porte"
"Moto Yamaha R1 da 1000cc"

```python
class Veicolo:
    def __init__(self, marca, modello):
        self.marca = marca
        self.modello = modello
    
    def descrizione(self):
        return f'{self.marca} {self.modello}'
    
class Auto(Veicolo):
    def __init__(self, marca, modello, porte):
        super().__init__(marca, modello)
        self.porte = porte
        
    def descrizione(self):
        return f'Auto {super().descrizione()} con {self.porte} porte'
    
class Moto(Veicolo):
    def __init__(self, marca, modello, cilindrata):
        super().__init__(marca, modello)
        self.cilindrata = cilindrata
        
    def descrizione(self):
        return f'Moto {super().descrizione()} da {self.cilindrata}cc'
    
a = Auto('Fiat', 'Panda', 5)
m = Moto('Yamaha', 'R1', 1000)

print(a.descrizione())
print(m.descrizione())
```

Crea una classe Evento con:
attributi: titolo, data_inizio, data_fine
usa il modulo datetime
Implementa:
metodo durata() → restituisce la durata in giorni
__str__ → rappresentazione leggibile
__lt__ → confronto tra eventi per data
Crea una lista di eventi e:
ordinali per data
stampa quelli futuri rispetto ad oggi

Vincoli:
Usare datetime e timedelta
Gestire correttamente confronti tra date

Output atteso:
Evento: Meeting (2026-04-01 → 2026-04-03)
Durata: 2 giorni

```python
from datetime import datetime

class Evento:
    
    def __init__(self, titolo, data_inizio, data_fine):
        self.titolo = titolo
        
        # Conversione di una data in formato stringa in datetime
        self.data_inizio = datetime.strptime(data_inizio, "%Y-%m-%d")
        self.data_fine = datetime.strptime(data_fine, "%Y-%m-%d")
        
        if self.data_fine < self.data_inizio:
            print('La data di fine non puòò essere precedente alla data di inizio')
        
    def durata(self):
        return (self.data_fine - self.data_inizio).days + 1
    
    def __str__(self):
        return f'Evento: {self.titolo} ({self.data_inizio.date()} → {self.data_fine.date()})'
    
    def __lt__(self, other):
        return self.data_inizio < other.data_inizio
    
# e = Evento('Meeting', '2026-04-01', '2026-04-03')
# print(e)
# print(f'Durata: {e.durata()} giorni')

eventi = [
    Evento('Meeting', '2026-06-01', '2026-06-03'),
    Evento('Workshop', '2026-02-27', '2026-03-02'),
    Evento('Conferenza', '2026-01-01', '2026-01-01')
]

eventi_odinati = sorted(eventi)

for evento in eventi_odinati:
    print(evento)
    print(f'Durata: {evento.durata()} giorni')
    
print('###################################')    
    
today = datetime.now()

for evento in eventi_odinati:
    if evento.data_inizio > today:
        print(evento)
        print(f'Durata: {evento.durata()} giorni')
```
