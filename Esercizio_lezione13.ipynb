Sistema di prenotazione corsi online con gestione completa delle eccezioni

Progettare un sistema per la gestione delle iscrizioni a corsi online.
Il sistema deve essere composto da una classe principale chiamata CorsoOnline che permetta di:
-> creare un corso
-> iscrivere studenti
-> rimuovere studenti
-> controllare posti disponibili
-> gestire tutti i possibili errori tramite eccezioni

CorsoOnline:
Attributi privati
    __nome
    __max_studenti
    __studenti → lista studenti iscritti

Metodo astratto
    inserisci un metodo descrizione_corso()
    che andrà implementato nelle sottoclassi

Metodi pubblici richiesti
    iscrivi_studente(nome_studente)
    Deve generare eccezione se:
        il nome è vuoto
        lo studente è già iscritto
        il corso è pieno

    rimuovi_studente(nome_studente)
    Deve generare eccezione se:
        il nome è vuoto
        lo studente non esiste

    posti_disponibili()
        Restituisce il numero di posti liberi

    lista_studenti()
        Restituisce una copia della lista studenti

    Eccezioni personalizzate obbligatorie
    Creare almeno 3 eccezioni personalizzate:
        class CorsoPienoError(Exception):
        class StudenteDuplicatoError(Exception):
        class StudenteNonTrovatoError(Exception):

Metodi dunder richiesti

Implementare:
__init__
__str__
__len__

Utilizzare ereditarietà per creare almeno 3 classi che ereditano da corsoOnline
Creare una lista di corsi utilizzando il polimorfismo
Utilizzare il concetto di astrazione e incapsulamento
Implementare un sistema di logging con la stampa in console e su file degli errori


```python
from abc import ABC, abstractmethod
import logging

# CONFIGURAZIONE LOGGING
logging.basicConfig(
    level = logging.ERROR,
    format = "%(asctime)s - %(levelname)s - %(message)s",
    handlers= [
        logging.FileHandler('errors.log'),
        logging.StreamHandler()
    ]
)

# ECCEZIONI PERSONALIZZATE
class CorsoPienoError(Exception): pass
class StudenteDuplicatoError(Exception): pass
class StudenteNonTrovatoError(Exception): pass

# CLASSE ASTRATTA
class CorsoOnline(ABC):
    def __init__(self, nome, max_studenti):
        
        if not isinstance(nome, str):
            raise TypeError('Il nome del corso non è valido')
        
        if not isinstance(max_studenti, int) or max_studenti <= 0:
            raise ValueError('Numero massimo di studenti non valido')
        
        self.__nome = nome
        self.__max_studenti = max_studenti
        self.__studenti = []
        
    # PROPERTY GETTER E SETTER
    @property
    def nome(self): return self.__nome
    
    # METODI DUNDER
    def __str__(self):
        return (f'Corso {self.__nome} | Studenti iscritti {len(self.__studenti)}/{self.__max_studenti}')
    
    def __len__(self):
        return len(self.__studenti)
    
    # METODI ASTRATTI
    @abstractmethod
    def descrizione_corso(self):
        pass
   
    # METODI PUBBLICI 
    def iscrivi_studente(self, nome_studente):
        if not isinstance(nome_studente, str):
            raise TypeError('Il nome dello studente non è valido')
        if not nome_studente.strip():
            raise ValueError('Nome studente vuoto')
        if nome_studente in self.__studenti:
            raise StudenteDuplicatoError(f'Lo studente {nome_studente} è già iscritto')
        if len(self.__studenti) >= self.__max_studenti:
            raise CorsoPienoError(f'Il corso {self.__nome} è pieno')
        
        self.__studenti.append(nome_studente)
        print(f'Studente {nome_studente} iscritto al corso {self.__nome}')
    
    def rimuovi_studente(self, nome_studente):
        if not isinstance(nome_studente, str):
            raise TypeError('Il nome dello studente non è valido')
        if not nome_studente.strip():
            raise ValueError('Nome studente vuoto')
        if nome_studente not in self.__studenti:
            raise StudenteNonTrovatoError(f'Lo studente {nome_studente} non trovato')
        
        self.__studenti.remove(nome_studente)
        print(f'Studente {nome_studente} rimosso dal corso {self.__nome}')
        
    def posti_disponibili(self):
        return self.__max_studenti - len(self.__studenti)
    
    def lista_studenti(self):
        return self.__studenti.copy()
    

# SOTTOCLASSI

class CorsoPython(CorsoOnline):   
    def descrizione_corso(self):
        return "Corso dedicato alla programmazione in Python"
    
class CorsoJava(CorsoOnline):
    def descrizione_corso(self):
        return "Corso dedicato alla programmazione in Java"

class CorsoSQL(CorsoOnline):
    def descrizione_corso(self):
        return "Corso dedicato alla gestione di Database"
    

# LISTA CORSI UTILIZZANDO IL POLIMORFISMO
lista_corsi = [
    CorsoPython('Corso Python base', 5),
    CorsoJava('Corso Java avanzato', 3),
    CorsoSQL('Corso MySQL completo', 6)
]
    
# TEST DEL SISTEMA
try:
    for corso in lista_corsi:
        print(corso)
        print(corso.descrizione_corso())
        print('Posti disponibili: ', corso.posti_disponibili())
    
    
    corso_python = lista_corsi[0]
    
    corso_python.iscrivi_studente('Mario Rossi')
    # corso_python.iscrivi_studente(25)
    # corso_python.iscrivi_studente(" ")
    # corso_python.iscrivi_studente('Mario Rossi')
    corso_python.iscrivi_studente('Giuseppe Verdi')
    corso_python.iscrivi_studente('Francesca Neri')
    corso_python.iscrivi_studente('Antonio Bianchi')
    
    corso_python.rimuovi_studente('Giuseppe Verdi')
    # corso_python.rimuovi_studente('Giuseppe Verdi')
    
    print(corso_python.posti_disponibili())
    
    print(corso_python)
    
    for studente in corso_python.lista_studenti():
        print(studente)
    
except CorsoPienoError as err:
    # print('Errore', err)
    logging.error(err)
except StudenteDuplicatoError as err:
    # print('Errore', err)
    logging.error(err)
except StudenteNonTrovatoError as err:
    # print('Errore', err)
    logging.error(err)
except ValueError as err:
    # print('Errore', err)
    logging.error(err)
except TypeError as err:
    # print('Errore', err)
    logging.error(err)
finally:
    print('Gestione iscrizione completata')
```
