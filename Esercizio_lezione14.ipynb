# Scrivere un programma Python che gestisca una rubrica di contatti 
# (nome, cognome, email, telefono), permettendo all’utente di 
# aggiungere nuovi contatti, salvarli su file e leggerli successivamente, 
# con un’attenta gestione delle eccezioni in caso di errori.
#
# Tramite una interfaccia da terminale L’utente può scegliere una 
# delle seguenti operazioni:
#   -> Aggiungi un nuovo contatto
#   -> Visualizza i contatti salvati
#   -> Esci
#
# Aggiunta contatto
#   Richiedere all’utente: nome, cognome, email, telefono.
#   Validare i dati:
#       Email deve contenere @.
#       Telefono deve essere numerico.
#       Nome e Cognome devono essere di tipo str e non vuoti
#       Gestire eventuali errori di input con try/except.
# Scrittura su file
#   Salvare i dati su un file di testo (rubrica.txt) o CSV.
#   Ogni contatto è salvato su una riga, separato da ,
# Lettura da file
#   Quando l’utente sceglie “Visualizza contatti”, leggere il file 
#       e mostrare i contatti formattati.
#   Gestire eccezioni:
#       File non trovato (FileNotFoundError)
#       File vuoto
#       Errore nella decodifica
# Gestione delle eccezioni
#   Utilizzare blocchi try/except per:
#       Errori nell’apertura/scrittura/lettura del file
#       Input non valido (email, nome, cognome, telefono)
#       Eccezioni generiche
#       Utilizzare finally per stampare un messaggio finale e fare chisure
#       in ogni caso (es. “Operazione completata”).
# Esempio di rubrica.txt:
#   Mario,Rossi,mario.rossi@gmail.com,3281234567 

```python

import os

def menu():
    print('1 - Aggiungi un nuovo contatto')
    print('2 - Visualizza i contatti salvati')
    print('0 - Esci')
    return int(input('Scegli un valore compreso tra 0, 1,2'))

def aggiungi_contatto():
    # Richiedere all’utente: nome, cognome, email, telefono.
    try :
        nome = input('Inserisci un nome valido')
        cognome = input('Inserisci un cognome valido')
        if not isinstance(nome, str) or not isinstance(cognome, str) :
            raise TypeError('Nome/Cognome dello studente deve essere di tipo testo')
        if not nome.isalpha() or not cognome.isalpha(): 
            raise TypeError('Nome/Cognome dello studente deve contenere solo caratteri')
        if not nome.strip() or not cognome.strip():
            raise ValueError('Nome/Cognome studente vuoto')
        email = input('Inserisci una email valida (deve contenere @)')
        if not isinstance(email, str):
            raise TypeError('Email deve essere di tipo testo')
        if '@' not in email:
            raise ValueError('L\'email non è valida')
        telefono = input('Inserisci un numero di telefono (deve contenere solo numeri)')
        if not telefono.isdigit():
            raise ValueError('Numero di telefono deve contenere solo numeri')
        
        f = open('rubrica.txt', 'a', encoding="utf-8")
        row = f"{nome},{cognome},{email},{telefono}\n"
        f.write(row)
        f.close()
    except TypeError as e:
        print(e)
    except ValueError as e:
        print(e)
    except Exception as e:
        print(e)
    else:
        print('Operazione di inserimento completata')
    finally:
        print('FINE')
    
def visualizza_contatti():
    try:
        if not os.path.exists('rubrica.txt'):
            raise FileNotFoundError('Il file \'rubrica.txt\' non esiste!')
        
        f = open('rubrica.txt', 'r', encoding="utf-8")
        contatti = f.readlines()
        if not contatti:
            raise ValueError('Nessun contatto salvato in rubrica')
        
        print()
        print('Lista Contatti')
        for i, contatto in enumerate(contatti, 1):
            dati = contatto.strip().split(',')
            nome, cognome, email, telefono = dati
            print(f'{i} - {nome} {cognome} Email: {email} Telefono: {telefono}')
        print()
    except ValueError as e:
        print(e)
    except FileNotFoundError as e:
        print(e)
    except Exception as e:
        print(e)
    else:
        print('Operazione di lettura completata')
        f.close()
    finally:
        print('FINE')
    
while True:
    scelta = menu()
    if scelta == 1:
        aggiungi_contatto()
    elif scelta == 2:
        visualizza_contatti()
    elif scelta == 0:
        print('Esci')
        break
    else:
        print('Hai inserito un valore errato')
```
