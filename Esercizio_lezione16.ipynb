Sistema Gestione Analisi Vendite

Una piccola azienda vuole analizzare le vendite dei propri prodotti.

Ogni vendita contiene:
- nome prodotto
- categoria
- quantità venduta
- prezzo unitario
- data vendita

Gli studenti dovranno sviluppare un programma Python che:
- memorizzi le vendite
- le salvi su MongoDB
- consenta alcune analisi statistiche semplici

Struttura richiesta del progetto
progetto_vendite/
│
├── main.py
├── models/
│   └── vendita.py
├── services/
│   └── database_service.py
└── utils/
    └── statistiche.py

Creare una classe chiamata Vendita.
Attributi richiesti
- prodotto
- categoria
- quantita
- prezzo_unitario
- data_vendita

Metodi richiesti
- Metodo totale() -> quantita * prezzo_unitario

Creare almeno le seguenti funzioni:

- crea_vendita()
- mostra_vendite(vendite)
- calcola_fatturato_totale(vendite)
- raggruppa_per_categoria(vendite)

Operazioni richieste
- Estrarre categorie uniche
Ordinare vendite
Ordinare per:
- prezzo
- oppure
- quantità

Gestione Eccezioni
Gestire correttamente:
- valori numerici non validi
- quantità negative
- prezzo negativo
- errori di connessione MongoDB

Analisi dati
Implementare le seguenti analisi:
- Prodotto con fatturato più alto
- Categoria più venduta
- Calcolare il prezzo medio dei prodotti.
