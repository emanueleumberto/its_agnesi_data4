Utilizzando il dataset SerieA_Season2026.csv, analizzare le statistiche delle squadre della Serie A attraverso operazioni di raggruppamento (groupby), aggregazione e ordinamento dei dati con la libreria Pandas.

Sei un Data Analyst che lavora per una redazione sportiva. Il tuo compito è estrarre statistiche significative dal campionato di Serie A 2025/2026 per creare report e classifiche.

Il dataset contiene informazioni sulle 380 partite della stagione, tra cui:

Squadra di casa (HomeTeam)
Squadra ospite (AwayTeam)
Gol segnati in casa (FTHG)
Gol segnati in trasferta (FTAG)
Risultato finale (FTR)
Tiri effettuati (HS, AS)
Tiri in porta (HST, AST)
Falli (HF, AF)
Cartellini gialli (HY, AY)
Cartellini rossi (HR, AR)
