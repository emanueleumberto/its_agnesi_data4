Sistema di Calcolo Sconto Cliente
Realizza un programma Python che simuli un sistema di calcolo sconto per un negozio.
Il programma deve:
- Richiedere all’utente:
    nome cliente
    importo totale della spesa
    età del cliente

- Applica uno sconto in base alle seguenti condizioni:
Se l’importo è:
    ≥ 100€ → sconto 10%
    ≥ 50€ e < 100€ → sconto 5%
    < 50€ → nessuno sconto
Sconto aggiuntivo:
    Se età < 18 → +5%
    Se età > 65 → +7%

Gli sconti sono cumulativi (somma delle percentuali)

Calcolare:
sconto totale (%)
valore sconto (€)
prezzo finale

Stampare un riepilogo con:
- Nome cliente
- Importo iniziale
- Percentuale sconto totale
- Importo sconto
- Prezzo finale
