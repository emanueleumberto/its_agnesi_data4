Sistema di Calcolo Sconto Cliente
Realizza un programma Python che simuli un sistema di calcolo sconto per un negozio.
Il programma deve:
- Richiedere all’utente:
    nome cliente
    importo totale della spesa
    età del cliente

- Applica uno sconto in base alle seguenti condizioni:
Se l’importo è:
    ≥ 100€ → sconto 10%
    ≥ 50€ e < 100€ → sconto 5%
    < 50€ → nessuno sconto
Sconto aggiuntivo:
    Se età < 18 → +5%
    Se età > 65 → +7%
Gli sconti sono cumulativi (somma delle percentuali)

Calcolare: sconto totale (%) - valore sconto (€) - prezzo finale

Stampare un riepilogo con:
- Nome cliente
- Importo iniziale
- Percentuale sconto totale
- Importo sconto
- Prezzo finale

```python
nome = input("Nome cliente:").strip()
importo= float(input("Importo totale:").strip())
eta = int(input("Età del cliente: ").strip())

# print(nome, importo, eta)
# print(type(nome), type(importo), type(eta))

sconto = 0
if importo >= 100:
    sconto += 10
elif importo >= 50 and importo < 100:
    sconto += 5
else:
    sconto += 0
    
if eta < 18:
    sconto += 5
elif eta > 65:
    sconto += 7
else:
    sconto += 0
    
pecentuale_sconto = f"{sconto}%"
# print(pecentuale_sconto)
valore_sconto = importo * (sconto / 100)
# print(valore_sconto)
sconto_totale = importo - valore_sconto
# print(sconto_totale)

print(f"Nome cliente: {nome}")
print(f"Importo iniziale: {importo}")
print(f"Percentuale sconto: {pecentuale_sconto}")
print(f"Importo sconto: {valore_sconto}")
print(f"Prezzo finale: {sconto_totale}")
```

Sistema di Accesso con Tentativi Limitati
Realizza un programma Python che simuli un sistema di login da terminale.

Il programma deve:
Richiedere all’utente: username e password
Confrontare i dati inseriti con credenziali predefinite:
username_correto = "admin"
password_corretta = "1234"

L’utente ha massimo 3 tentativi
Ad ogni errore:
 - mostrare un messaggio di errore
 - indicare i tentativi rimanenti
Se le credenziali sono corrette:
 - stampare messaggio di accesso consentito
 - terminare il ciclo
Se i tentativi finiscono:
 - bloccare l’accesso
 - stampare messaggio finale

Controlli obbligatori
Username vuoto → messaggio di errore
Password vuota → messaggio di errore

Output atteso (esempio)
Tentativo 1/3
Username: admin
Password: 1111
Credenziali errate. Tentativi rimasti: 2

Tentativo 2/3
Username: admin
Password: 1234
Accesso consentito!
