Sistema di Gestione Voti Studenti
Realizza un programma Python che permetta di:

Inserire una serie di voti (numeri interi)
Analizzare i voti inseriti
Mostrare un riepilogo finale

Inserimento dati
- Chiedere all’utente di inserire voti uno alla volta da terminale
- Salvare i voti in una lista
- L’inserimento termina quando l’utente digita: fine

Durante l’inserimento:
Il voto deve essere:
- numerico
- compreso tra 0 e 30
In caso di errore:
- mostrare un messaggio
- NON aggiungere il valore alla lista

Elaborazione dati
Utilizzando un ciclo for, calcolare:
 - Somma totale dei voti
 - Media dei voti
 - Numero di voti sufficienti (≥ 18)
 - Numero di voti insufficienti (< 18)
 - Voto massimo
 - Voto minimo

Usare if-elif-else per classificare la media:
- Media ≥ 27 → "Eccellente"
- Media ≥ 21 → "Buono"
- Media ≥ 18 → "Sufficiente"
- Media < 18 → "Insufficiente"

Output richiesto

Stampare:
- Lista voti inseriti
- Media
- Numero sufficienti / insufficienti
- Voto massimo e minimo
- Classificazione finale

```python
voti = []

while True:
    # Inserire una serie di voti (numeri interi)
    valore = input('Inserisci un voto e digita: fine per terminare: ')

    # Controllo se il ciclo deve terminare
    if valore.lower() == 'fine':
        print("Fine inserimento.")
        break
    
    # Durante l’inserimento:
    # Il voto deve essere:
    # numerico
    if valore.isdigit():
        voto = int(valore)
        # compreso tra 0 e 30
        if voto >= 0 and  voto <= 30:
            voti.append(voto)
        else:
            print("Devi inserire un valore compreso tra 0 e 30")
    else:
        print("Devi inserire un valore numerico.")

print("Inserimento voti teminato.")

# Analizzare i voti inseriti
if len(voti) == 0:
    print("Nessun voto inserito")
else:
    somma = 0 # Somma totale dei voti
    sufficienti = 0 # Numero di voti sufficienti (≥ 18)
    insufficienti = 0 # Numero di voti insufficienti (< 18)
    massimo = voti[0] # Voto massimo
    minimo = voti[0] # Voto minimo
    
    for v in voti:
        somma += v # Somma dei voti
        
        if v >= 18: sufficienti += 1 # Conto voti sufficienti 
        else: insufficienti +=1 # Conto voti insufficienti 
        
        if v > massimo: massimo = v # Calcolo il Voto massimo
        if v < minimo: minimo = v # Calcolo il Voto minimo
        
        
    media = somma / len(voti) # Media dei voti 
    
# classificare la media
classificazione = ""

if media >= 27: classificazione = "Eccellente"
elif media >= 21: classificazione = "Buono"
elif media >= 18: classificazione = "Sufficiente"
else: classificazione = "Insufficiente"

# Mostrare un riepilogo finale
# Output richiesto
# Stampare:
print(voti) # Lista voti inseriti
print("Media voti: ", media) # Media voti
print("Sufficienti: ", sufficienti,'/', insufficienti) # Numero sufficienti / insufficienti
print("Voto Massimo: ", massimo) # Voto massimo
print("Voto Minimo: ", minimo) # Voto minimo
print("Classificazione: ", classificazione) # Classificazione finale


```
