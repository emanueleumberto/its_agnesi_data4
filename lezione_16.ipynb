```python
from pymongo import MongoClient

cluster = MongoClient('mongodb+srv://admin:Qp1CxtD4s3j6q6TB@cluster0.7opl1ou.mongodb.net/?appName=Cluster0')
db = cluster.DataMongoPython3
users = db['Users']

# Modificare un dato nella collection MongoDB -> Update in SQL
# update_one(filter, update)
# update_many(filter, update)

# il primo parametro del metodo update_one() è un oggetto di tipo query
# che definisce quale elemento aggiornate, il secondo parametro è
# è un oggetto con le modifiche da applicare all'elemento selezionato

result = users.update_one({'city': 'Milano'}, {'$set': {'firstname': 'Luigi', 'age': 99}})
print(result.modified_count, 'Modified!!')

# il primo parametro del metodo update_many() è un oggetto di tipo query
# che definisce quale elemento aggiornate, il secondo parametro è
# è un oggetto con le modifiche da applicare a tutti gli elementi selezionato

result = users.update_many({'city': 'Milano'}, {'$set': {'firstname': 'Luigi', 'age': 99}})
print(result.modified_count, 'Modified!!')

cluster.close()
```

```python
from pymongo import MongoClient

cluster = MongoClient('mongodb+srv://admin:Qp1CxtD4s3j6q6TB@cluster0.7opl1ou.mongodb.net/?appName=Cluster0')
db = cluster.DataMongoPython3
users = db['Users']

# Eliminare un dato da una collection -> DELETE in SQL
# delete_one(filter)
# delete_many(filter)

# Il parametro del metodo delete_one è un oggetto di tipo query
# che definisce quale elemento eliminare nella collection
result = users.delete_one({'city': 'Napoli'})
print(result.deleted_count, 'deleted!')

# Il parametro del metodo delete_many è un oggetto di tipo query
# che definisce tutti gli elementi da eliminare nella collection
result = users.delete_many({'city': 'Napoli'})
print(result.deleted_count, 'deleted!')

cluster.close()
```
