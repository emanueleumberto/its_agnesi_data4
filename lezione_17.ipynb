```python
import numpy as np

arr = np.array([26,65,31,79,22])

# Filtrare un array tramite i metodi di numpy
arrFilter = np.where(arr > 40) # Restituisce l'indice degli elementi che risponde alla condizione
# print(arrFilter)
print(arr[arrFilter])

arrSorted = np.sort(arr) # Ordina in ASC
print(arrSorted)
arrSortedReverse = -np.sort(-arr) # Ordina in DESC
print(arrSortedReverse)

# Valori Random tramite numpy
rand = np.random.randint(50, 100) # Genera un valore random da 50 a 100
rand = np.random.rand() # Genera un valore random di tipo Float da 0 a 1
print(rand)

arrRand = np.random.rand(5,2) # Genera un array multidimensionale con valori random
arrRand = np.random.randint(50, 100, size=10) # Genera un array monodimensionale con valori random da 50 a 100
print(arrRand)

arrChoice = np.random.choice(arrRand) # Sceglie un valore casuale tra tutti i valori presenti in un array che gli fornisco come paramentro
arrChoice = np.random.choice(arrRand, size=3) # Scegli size valori casuali tra quelli che gli fornisco come paramentro
print(arrChoice)
```

```python
# Pandas

# Pandas introduce due nuovi tipi di dato
# Series e DataFrame
# Series -> prendono in ingresso una lista di valori
# Dataframe -> prendono in ingresso un dizionario o un ndArray
# Dataframe è un tipo di dato ottimizzato per lavorare con tabelle bidimensionali
# come quelle che utilizziamo su un CSV o in Excel o tramite un JSON o dati di tipo SQL

# pip install pandas
```

```python
import numpy as np
import pandas as pd

# print("Versione NumPy: ", np.__version__)
# print("Versione Pandas: ", pd.__version__)

# Series e DataFrame

# Series
# prende in ingresso una lista di valori

pds = pd.Series([26,65,31,79,22])
# print(pds)
# print(pds[2])
# print(pds[1:3])

pdsi = pd.Series([26,65,31,79,22], index=list(('abc', 'cdv', 'tdr', 'poi', 'sus')))
# print(pdsi)
# print(pdsi['cdv'])
# print(pdsi['cdv':'poi'])


# Esistono due metodi diversi nelle Series che permettono
# l'accesso a un sottoinsieme di dati
# loc[] -> usa le etichette
# iloc[] -> usa gli indici (slicing)

pdsi = pd.Series([26,65,31,79,22], index = ['2000', '2001', '2002', '2003', '2004'])
print(pdsi)
print(pdsi.iloc[2])
print(pdsi.loc['2002'])
print(pdsi.iloc[1:3])
print(pdsi.loc['2001':'2003'])
```

```python
# DataFrame
# Dataframe è un tipo di dato ottimizzato per lavorare con tabelle bidimensionali
# come quelle che utilizziamo su un CSV,in Excel, tramite un JSON o dati di tipo SQL
# Dataframe -> prendono in ingresso un dizionario o un ndArray

import numpy as np
import pandas as pd

arr = np.random.randint(50, 100, [4,5])
# data = pd.DataFrame(arr)

# Posso esplicitare intestazioni di riga e/o di colonna nella
# rappresentazione del dato di un DataFrame
data = pd.DataFrame(arr, 
                    index=['Roma', 'Milano', 'Torino', 'Napoli'],
                    columns=['2000', '2001', '2002', '2003', '2004'])
# print(data)


# Dizionario è un dato composto da una serie di coppie chiave/valore
obj = {
    'Data Analyst': [10, 20, 30],
    'Cyber Security': [40, 50, 60],
    'Developer': [70, 80, 90],
    'Cloud Dev': [10, 40, 70]
}
# print(obj)
data = pd.DataFrame(obj)
# Posso esplicitare intestazioni di riga per la
# rappresentazione del dato di un DataFrame
# le intestazioni di colonna vengono prese di default dalle chiavi del dizionario
data = pd.DataFrame(obj, index=['Roma', 'Milano', 'Torino'])
data
```
