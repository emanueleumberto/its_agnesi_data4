```python
import numpy as np
import pandas as pd

# print("Versione NumPy: ", np.__version__)
# print("Versione Pandas: ", pd.__version__)

# Series e DataFrame

# Series
# prende in ingresso una lista di valori

pds = pd.Series([26,65,31,79,22])
# print(pds)
# print(pds[2])
# print(pds[1:3])

pdsi = pd.Series([26,65,31,79,22], index=list(('abc', 'cdv', 'tdr', 'poi', 'sus')))
# print(pdsi)
# print(pdsi['cdv'])
# print(pdsi['cdv':'poi'])


# Esistono due metodi diversi nelle Series che permettono
# l'accesso a un sottoinsieme di dati
# loc[] -> usa le etichette
# iloc[] -> usa gli indici (slicing)

pdsi = pd.Series([26,65,31,79,22], index = ['2000', '2001', '2002', '2003', '2004'])
print(pdsi)
print(pdsi.iloc[2])
print(pdsi.loc['2002'])
print(pdsi.iloc[1:3])
print(pdsi.loc['2001':'2003'])
```

```python
import pandas as pd

file_path = 'irisNull.csv'

data = pd.read_csv(file_path, delimiter=',', encoding='ISO-8859-1')

print(data.shape)
# data.info()

# metodi predefiniti per la pulizia dei dati 
# isnull()
# dropna()

data.isnull() # restituisce un dataframe con valori booleani sulla presenza di campi null 
data.isnull().sum() # restituisce un riepilogo di valori di tipo null per ogni colonna

data.dropna() # elimino tutte le righe con valori mancanti
data.dropna(axis=0) # elimino tutte le righe con valori mancanti
data.dropna(axis=1) # elimino tutte le colonne con valori mancanti
data.dropna(how='any') # elimina una riga se c'è almeno un valore mancante 
data.dropna(how='all') # elimina una riga se tutti i valori sono mancanti
data.dropna(thresh=2) # elimina una riga se ci sono N valori mancanti 

data.fillna(10) # Sostituisce i valori mancanto con il valore fisso, passato come paramentro
data.fillna({'variety': 'Sconsciuta', 'petal.length': 99}) # Sostituisce solo nelle colonne passate come parametro i valori Nan
data.ffill() # Sostituisci i dati mancanti con quelli della riga precedente
data.bfill() # Sostituisci i dati mancanti con quelli della riga successiva
data.fillna(data.mean(numeric_only=True)) # Sostituisci i dati mancanti con la media dei valori della colonna numerica

data[data.duplicated()] # Verifica e restituisce dati duplicati nel dataframe
data[data.duplicated(subset=['variety'])] # Verifica e restituisce dati duplicati in una colonna specifica nel dataframe

data.drop_duplicates() # Elimina dati duplicati nel dataframe
data.drop_duplicates(subset=['variety']) # Elimina dati duplicati in una colonna specifica nel dataframe
```

```python
import pandas as pd

ldata = { 'c': ['uno', 'due', 'tre', 'quattro', 'cinque'], 'v': [1,2,3,4,5]}
rdata = { 'c': ['uno', 'sei', 'due', 'sette', 'otto'], 'v': [6,7,8,9,10]}

leftData = pd.DataFrame(ldata)
rightData = pd.DataFrame(rdata)

pd.merge(left=leftData, right=rightData, on='c', how='inner')
pd.merge(left=leftData, right=rightData, on='c', how='left')
pd.merge(left=leftData, right=rightData, on='c', how='right')
pd.merge(left=leftData, right=rightData, on='c', how='outer')
```
